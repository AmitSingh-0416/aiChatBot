# Task
Create a LangGraph workflow with at least 3 nodes to handle customer queries, use memory for context, and provide predictive suggestions based on mock customer intelligence logic.

```python
from typing import TypedDict

class GraphState(TypedDict):

    query: str
    context: str
    customer_profile: str
    suggestion: str
```

## Workflow nodes

### Subtask:
Three nodes:
- Node 1: Capture the customer query and retrieve relevant context (mock retrieval).
- Node 2: Run basic customer segmentation logic based on the query and context to create a mock customer profile.
- Node 3: Based on the mock customer profile, suggest a product, offer, or response.


**Reasoning**:
Define the three functions for the nodes of the LangGraph workflow as described in the instructions.



```python
def capture_query_and_retrieve_context(state):
    """
    Node 1
    """
    print("---NODE 1: CAPTURE QUERY & RETRIEVE CONTEXT---")
    query = state.get("query", "")
    context = f"Simulated context for query: '{query}'. Customer is interested in pricing and features."
    return {"context": context}

def run_customer_segmentation(state):
    """
    Node 2
    """
    print("---NODE 2: RUN CUSTOMER SEGMENTATION---")
    query = state.get("query", "")
    context = state.get("context", "")
    customer_profile = "Standard Customer"
    if "pricing" in query.lower() or "cost" in query.lower() or "pricing" in context.lower():
        customer_profile = "Price-Sensitive Customer"
    elif "features" in query.lower() or "capabilities" in query.lower() or "features" in context.lower():
        customer_profile = "Feature-Focused Customer"
    return {"customer_profile": customer_profile}

def suggest_response(state):
   """
   Node 3
   """
    print("---NODE 3: SUGGEST RESPONSE---")
    customer_profile = state.get("customer_profile", "")
    suggestion = "Thank you for your inquiry."
    if customer_profile == "Price-Sensitive Customer":
        suggestion = "Based on your interest in pricing, we recommend checking our current offers and pricing plans."
    elif customer_profile == "Feature-Focused Customer":
        suggestion = "Based on your interest in features, we recommend exploring our detailed features page or requesting a demo."
    return {"suggestion": suggestion}
```

## Build the langgraph

### Subtask:
Connect the nodes to build the LangGraph workflow.

```python
from langgraph.graph import StateGraph

# Create a StateGraph instance
workflow = StateGraph(GraphState)

# Add nodes to the graph
workflow.add_node("retrieve_context", capture_query_and_retrieve_context)
workflow.add_node("segment_customer", run_customer_segmentation)
workflow.add_node("suggest_response", suggest_response)

# Set the entry point
workflow.set_entry_point("retrieve_context")

# Add edges
workflow.add_edge("retrieve_context", "segment_customer")
workflow.add_edge("segment_customer", "suggest_response")

# Set the finish point
workflow.set_finish_point("suggest_response")

# Compile the graph
app = workflow.compile()
```
