## Set up the language model and memory

```python
from langchain_community.chat_models import ChatOpenAI
from langchain.memory import ConversationBufferMemory
import os

# Set your OpenAI API key here
os.environ["OPENAI_API_KEY"] = "MY_API_KEY" # Replace with your actual API key

# Instantiate the language model
llm = ChatOpenAI(temperature=0)

# Instantiate the memory module
memory = ConversationBufferMemory(memory_key="chat_history")

print("Language model and memory module initialized.")
```


Import the necessary library for sentiment analysis and initialize a sentiment analysis pipeline.



```python
from transformers import pipeline

# Initialize the sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis")
```


Create a function to analyze sentiment using the initialized pipeline.



```python
def get_sentiment(text):

  result = sentiment_analyzer(text)
  return result[0]['label']

# Example usage:
print(get_sentiment("I love this product!"))
print(get_sentiment("This is terrible."))
```

## Create a custom prompt template


```python
from langchain.prompts import PromptTemplate

template = """You are a sentiment-aware chatbot. You should respond to the user based on the conversation history and the sentiment of their latest input.

Chat History:
{chat_history}
Human: {human_input}
Sentiment: {sentiment}
Chatbot:"""

prompt = PromptTemplate(
    input_variables=["chat_history", "human_input", "sentiment"],
    template=template
)

print("Prompt template created.")
```

## Build the chatbot chain



```python
from langchain.chains import LLMChain

conversation_chain = LLMChain(
    llm=llm,
    prompt=prompt,
    memory=memory
)

print("Conversation chain created.")
```

## Implement sentiment-aware responses



```python
from transformers import pipeline

# Initialize the sentiment analyzer
sentiment_analyzer = pipeline("sentiment-analysis")

def generate_sentiment_aware_response(user_input, conversation_chain):

  sentiment = get_sentiment(user_input)
  print(f"Detected sentiment: {sentiment}")
  response = conversation_chain.run(human_input=user_input, sentiment=sentiment)



  return response
```

```python

from langchain_community.chat_models import ChatOpenAI
from langchain.memory import ConversationBufferMemory
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from transformers import pipeline
import os


llm = ChatOpenAI(temperature=0)

# Re-create the conversation chain components

# Define the prompt template again.
template = """You are a sentiment-aware chatbot. You should respond to the user based on the conversation history and the sentiment of their latest input.

Chat History:
{chat_history}
Human: {human_input}
Sentiment: {sentiment}
Chatbot:"""

# Create the PromptTemplate instance.
prompt = PromptTemplate(
    input_variables=["chat_history", "human_input", "sentiment"],
    template=template
)

# Create the ConversationBufferMemory instance.
memory = ConversationBufferMemory(memory_key="chat_history")

# Create the LLMChain instance, combining the LLM, prompt, and memory.
conversation_chain = LLMChain(
    llm=llm,
    prompt=prompt,
    memory=memory
)

print("Language model, memory module, and conversation chain re-initialized.")

# Initialize the sentiment analyzer again for the test run.
sentiment_analyzer = pipeline("sentiment-analysis")

# Re-define the get_sentiment function.
def get_sentiment(text):

  result = sentiment_analyzer(text)
  return result[0]['label']

# Re-define the generate_sentiment_aware_response function.
def generate_sentiment_aware_response(user_input, conversation_chain):

  sentiment = get_sentiment(user_input)
  print(f"Detected sentiment: {sentiment}")

  response = conversation_chain.run(human_input=user_input, sentiment=sentiment)

  return response

# Define a list of test cases (user inputs).
test_cases = [
    "Hello, how are you?",
    "I am feeling great today!",
    "That's interesting. Tell me more.",
    "I had a really bad day at work.",
    "I am not sure about that.",
    "But I am trying to stay positive.",
    "This is the worst thing that ever happened.",
    "However, I believe things will get better.",
    "Okay, let's change the topic."
]

# Iterate through the test cases and get responses from the chatbot.
for user_input in test_cases:
  print(f"\nUser Input: {user_input}")
  # Call the response generation function for each input.
  response = generate_sentiment_aware_response(user_input, conversation_chain)
  print(f"Chatbot Response: {response}")
```
